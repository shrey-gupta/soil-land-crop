```python
# List of Midwest state acronyms
midwest_states = ['IL', 'IN', 'IA', 'KS', 'MI', 'MN', 'MO', 'ND', 'NE', 'OH', 'SD', 'WI']

# Rearrange the columns
columns_order = [
    'fips', 'name', 'state', 'day', 'month', 'year',  # First columns
    'PRECTOT', 'PS', 'QV2M', 'T2M', 'T2MDEW', 'T2MWET', 'T2M_MAX', 'T2M_MIN',
    'T2M_RANGE', 'TS', 'WS10M', 'WS10M_MAX', 'WS10M_MIN', 'WS10M_RANGE',
    'WS50M', 'WS50M_MAX', 'WS50M_MIN', 'WS50M_RANGE', 'lat', 'lon', 
    'elevation', 'WAT_LAND', 'NVG_LAND', 'URB_LAND', 'GRS_LAND', 'FOR_LAND',
    'CULTRF_LAND', 'CULTIR_LAND', 'CULT_LAND',  # Middle columns
    'score'  # Last column
]

####### For 2015
# Split into train and test based on Midwest states
test_df_15 = draught_df_15[draught_df_15['state'].isin(midwest_states)].reset_index(drop=True)
test_df_15 = test_df_15.drop(columns=['name', 'state'])

train_df_15 = draught_df_15[~draught_df_15['state'].isin(midwest_states)].reset_index(drop=True)
train_df_15 = train_df_15.drop(columns=['name', 'state'])


###### For all the years
# Split into train and test based on Midwest states
test_df = draught_mg_df[draught_mg_df['state'].isin(midwest_states)].reset_index(drop=True)
test_df = test_df.drop(columns=['name', 'state'])

train_df = draught_mg_df[~draught_mg_df['state'].isin(midwest_states)].reset_index(drop=True)
train_df = train_df.drop(columns=['name', 'state'])
```

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import MinMaxScaler
from xgboost import XGBClassifier
from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score

exclude_cols = ['fips', 'year', 'month', 'day', 'lat', 'lon', 'score']
features_to_normalize = train_df_15.columns.difference(exclude_cols)

X_train = train_df_15.drop(columns=['score'])  
y_train = train_df_15['score']

X_test = test_df_15.drop(columns=['score'])  
y_test = test_df_15['score']

## Normalize
scaler = MinMaxScaler()
X_train[features_to_normalize] = scaler.fit_transform(X_train[features_to_normalize])
X_test[features_to_normalize] = scaler.fit_transform(X_test[features_to_normalize])

# Define the XGBoost classifier
xgb_model = XGBClassifier(eval_metric='logloss', random_state=42) ##use_label_encoder=False,

# Define hyperparameter grid for tuning
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.1, 0.2]
    # 'subsample': [0.8, 1],
    # 'colsample_bytree': [0.8, 1]
}

# Perform hyperparameter tuning using GridSearchCV
grid_search = GridSearchCV(
    estimator = xgb_model,
    param_grid = param_grid,
    scoring = 'accuracy',
    cv=5,
    verbose=1,
    n_jobs=-1
)

# Fit the model to the training data
grid_search.fit(X_train, y_train)

# Get the best parameters
best_params = grid_search.best_params_
print(f"Best Parameters: {best_params}")

# Evaluate the model on the test set
best_xgb = grid_search.best_estimator_
y_pred = best_xgb.predict(X_test)

# Print classification report
print(classification_report(y_test, y_pred))

# xgb_model = XGBClassifier(eval_metric='logloss', n_estimators = 200, max_depth = 5, learning_rate = 0.01, random_state=42) ##use_label_encoder=False,
# xgb_model.fit(X_train, y_train)

# y_pred = xgb_model.predict(X_test)

# accuracy = accuracy_score(y_test, y_pred)
# print(f"Accuracy: {accuracy:.4f}")
```

```python
from matplotlib.animation import FuncAnimation

# Load the county shapefile
shapefile_path = "/Users/shrey/Downloads/Soil_Land_Crop/datasets/Shape-files/cb_2018_us_county_5m/cb_2018_us_county_5m.shp"
us_counties = gpd.read_file(shapefile_path)

# Filter for the conterminous US counties
exclude_states = ['02', '15', '72', '66', '78', '60', '69']  # FIPS codes for excluded states/territories
conterminous_us = us_counties[~us_counties['STATEFP'].isin(exclude_states)]

# Load the drought dataset (assuming it's named `drought_df` and contains 2015 data)
draught_df_2015 = train_df[train_df['year'] == 2015]

# Aggregate scores by county and month
draught_df_agg = draught_df_2015.groupby(['fips', 'month'], as_index=False).mean()

# Merge the drought data with the shapefile data
conterminous_us['fips'] = conterminous_us['GEOID'].astype(int)  # Ensure FIPS codes are integers
counties_with_draught = conterminous_us.merge(draught_df_agg, on='fips', how='left')

# Initialize the plot
fig, ax = plt.subplots(figsize=(14, 10))
ax.set_title("Drought Levels by County Over 2015", fontsize=16)
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")

# Initialize a colorbar placeholder
sm = plt.cm.ScalarMappable(cmap='YlOrRd', norm=plt.Normalize(vmin=0, vmax=5))
sm._A = []

# Function to update the plot for each month
def update(month):
    ax.clear()
    ax.set_title(f"Drought Levels by County - 2015, Month: {month}", fontsize=16)
    ax.set_xlabel("Longitude")
    ax.set_ylabel("Latitude")
    
    # Filter counties for the current month
    month_data = counties_with_draught[counties_with_draught['month'] == month]
    
    # Plot counties with drought scores
    month_data.plot(
        column='score',
        cmap='YlOrRd',
        edgecolor='white',
        legend=False,
        ax=ax
    )
    
    # Add the colorbar
    cbar = fig.colorbar(sm, ax=ax, orientation='horizontal', pad=0.01, aspect=40)
    cbar.set_label('Drought Level (Score)', fontsize=12)

# Create the animation
ani = FuncAnimation(
    fig, update, frames=range(1, 13), interval=1000, repeat=True
)

# Show the animation
plt.show()
```
